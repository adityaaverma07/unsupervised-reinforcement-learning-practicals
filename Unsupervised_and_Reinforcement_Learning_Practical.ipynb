# Part A: Customer Segmentation Using K-Means

## Business Problem

An online retailer wants to understand different types of customers.

The company has two simple pieces of information:

- Monthly Spending
- App Visits

We will use K-Means Clustering to divide customers into **3 groups**.

## Step 1: Import the Required Libraries

We will use:

- `pandas` to create and view the dataset
- `KMeans` to create customer groups
- `matplotlib` to visualize the clusters

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

## Step 2: Create a Small Customer Dataset

Each row represents one customer.

```python
data = {
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Monthly_Spending": [9000, 8500, 1200, 1500, 5000, 5500, 8800, 1800],
    "App_Visits": [20, 18, 3, 4, 10, 12, 19, 5]
}

df = pd.DataFrame(data)

df
```

### Observe the Data

Think about these questions:

1. Which customers seem to spend the most?
2. Which customers visit the app most often?
3. Can you already notice any natural groups?

## Step 3: Select the Features

For clustering, we will use:

- Monthly Spending
- App Visits

The customer name is only used as an identifier.

```python
X = df[["Monthly_Spending", "App_Visits"]]

X
```

## Step 4: Apply K-Means Clustering

We will ask the model to create:

**K = 3 clusters**

```python
model = KMeans(n_clusters=3, random_state=42, n_init=10)

df["Cluster"] = model.fit_predict(X)

df
```

## What Does the Cluster Column Mean?

The model has placed customers into 3 groups.

You may see cluster labels such as:

- Cluster 0
- Cluster 1
- Cluster 2

These numbers do **not** mean good, average or bad customers.

They are simply group labels created by the model.

## Step 5: Visualize the Customer Groups

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Monthly_Spending"],
    df["App_Visits"],
    c=df["Cluster"],
    s=100
)

for i, customer in enumerate(df["Customer"]):
    plt.text(
        df["Monthly_Spending"][i] + 100,
        df["App_Visits"][i],
        customer
    )

plt.xlabel("Monthly Spending")
plt.ylabel("App Visits")
plt.title("Customer Segmentation Using K-Means")
plt.show()
```

## Step 6: View Customers Cluster by Cluster

```python
df.sort_values("Cluster")
```

# Part B: Introduction to Reinforcement Learning

Reinforcement Learning is different from clustering.

In Reinforcement Learning, a system learns by:

**Taking an Action → Receiving a Reward → Learning from the Result**

We will understand this using a very simple delivery-route example.

## Business Scenario

A delivery company has two possible routes:

- Route A
- Route B

The company wants to choose the route that usually gives faster delivery.

For simplicity:

- Fast delivery gives a higher reward.
- Slow delivery gives a lower reward.

## Step 1: Define Some Example Rewards

```python
route_rewards = {
    "Route A": [5, 4, 6, 5, 4],
    "Route B": [8, 9, 7, 10, 8]
}

route_rewards
```

## Step 2: Calculate the Average Reward

```python
for route, rewards in route_rewards.items():
    average_reward = sum(rewards) / len(rewards)
    print(route, "Average Reward =", average_reward)
```

## Interpretation

If Route B has a higher average reward, the system may prefer Route B more often.

This represents the basic Reinforcement Learning idea:

- **Agent:** Delivery decision system
- **Environment:** Roads and traffic
- **Action:** Choose Route A or Route B
- **Reward:** Feedback based on delivery performance

## Exploration vs Exploitation

### Exploration
Try a new or less-used option.

Example:
Try Route A even if Route B usually performs better.

### Exploitation
Use the option that is already known to perform well.

Example:
Choose Route B because it has previously given higher rewards.

## Simple Exploration Example

The following code randomly chooses a route.

This represents **exploration**.

```python
import random

choice = random.choice(["Route A", "Route B"])

print("Exploration choice:", choice)
```

## Simple Exploitation Example

The following code chooses the route with the higher average reward.

This represents **exploitation**.

```python
average_A = sum(route_rewards["Route A"]) / len(route_rewards["Route A"])
average_B = sum(route_rewards["Route B"]) / len(route_rewards["Route B"])

if average_A > average_B:
    print("Best known route: Route A")
else:
    print("Best known route: Route B")
```
